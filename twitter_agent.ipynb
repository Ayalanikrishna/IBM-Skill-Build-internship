```python
model = genai.GenerativeModel('models/gemini-1.5-flash')
```

```python
def tweet_generator(prompt_text):
  tweet_prompt = (
      f"u r very talented person write a tweet (250 characters text)"
      f"based on this prompt{prompt_text}"
      f"keep it short,creative and use hashtags"
  )

  response = model.generate_content(tweet_prompt)
  tweet = response.text.strip()


  return tweet
```

```python
while True:
  prompt = input("enter the prompt for tweet: Exit to Quit")
  if prompt.lower() == "exit":print("exiting")
  break

  tweet = tweet_generator(prompt)
  print("genetrated tweet is",tweet ,"\n")
```
